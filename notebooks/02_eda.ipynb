```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
FIGURES = PROJECT_ROOT / 'reports' / 'figures'
FIGURES.mkdir(parents=True, exist_ok=True)
df = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'modeling_dataset.csv', parse_dates=['date'])
df.head()
```

```python
ax = df.plot(x='date', y='target', marker='o', figsize=(10, 5), title='Demand trend over time')
ax.set_ylabel('Taxi demand')
plt.tight_layout()
plt.savefig(FIGURES / 'eda_demand_trend.png', dpi=150)
plt.show()
```

```python
monthly = df.groupby('month')['target'].mean()
ax = monthly.plot(kind='bar', figsize=(10, 5), title='Average demand by month')
ax.set_ylabel('Average demand')
plt.tight_layout()
plt.savefig(FIGURES / 'eda_average_demand_by_month.png', dpi=150)
plt.show()
```

```python
ax = df.plot.scatter(x='monthly_total_rain', y='target', figsize=(7, 5), title='Rainfall vs taxi demand')
plt.tight_layout()
plt.savefig(FIGURES / 'eda_rainfall_vs_demand.png', dpi=150)
plt.show()
```

```python
corr = df.select_dtypes('number').corr()
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr.columns)), corr.columns, rotation=90)
ax.set_yticks(range(len(corr.columns)), corr.columns)
fig.colorbar(im)
plt.tight_layout()
plt.savefig(FIGURES / 'eda_correlation_heatmap.png', dpi=150)
plt.show()
```
